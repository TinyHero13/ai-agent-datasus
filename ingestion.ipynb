```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("encoding", "ISO-8859-1") \
    .option("delimiter", ";") \
    .load("s3a://ckan.saude.gov.br/SRAG/2025/INFLUD25-13-10-2025.csv")

df.display()
```

```python
%sql
CREATE CATALOG IF NOT EXISTS srag_datasus;
USE CATALOG srag_datasus;
CREATE SCHEMA IF NOT EXISTS raw;
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("srag_datasus.raw.srag_data")
```
